```python
import numpy as np
```

```python
arr = [
    [   # Layer 0
        [1, 2, 3],
        [4, 5, 6]
    ],
    [   # Layer 1
        [7, 8, 9],
        [10, 11, 12]
    ]
]
```

```python
# Accessing Elements
arr[0][1][2]   # Output: 6
```

```python
# Get a full layer
arr[0]
```

```python
# Get a specific row from a layer
arr[1][0]
```

```python
# Get a specific column from all rows in a layer
[col[1] for col in arr[0]]
```

```python
# Slice multiple layers
arr[0:2]
```

```python
# Slice rows inside a layer
arr[0][0:1]
```

```python
# Slice columns inside a row
arr[0][0][1:3]
```

```python
# 3D Array Slicing using NumPy
import numpy as np

a = np.array(arr)
# Get all layers, all rows, column 1
a[:, :, 1]
```

```python
# Get layer 1 completely
a[1, :, :]
```

```python
# Get first row from all layers
a[:, 0, :]
```

```python
# Slice layers & columns
a[0:2, :, 0:2]
```
