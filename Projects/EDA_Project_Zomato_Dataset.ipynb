## 8. Online Order vs Table Booking Analysis

### Objective
To examine the relationship between online ordering and table-booking availability across restaurants.

### Analysis
A cross-tabulation was created using the `pd.crosstab()` function to compare online-order availability with table-booking availability.


```python
#Online Order vs Table Booking
pd.crosstab(dataset['online_order'], dataset['table booking'])
```

```python
#Online Order vs Table Booking - Bar Plot Visualization
booking_comparision = pd.crosstab(dataset['online_order'],dataset['table booking'])
booking_comparision.plot(kind='bar', figsize=(7,5))
plt.title("Online Ordering vs Table Booking")
plt.xlabel("Online Order")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.legend(title='Table Booking')
plt.tight_layout()
plt.show()
```

```python
online_yes = dataset[dataset['online_order'] == 'Yes']

table_booking_yes = online_yes[
    online_yes['table booking'] == 'Yes'
]

percentage = len(table_booking_yes) / len(online_yes) * 100

print(f"Table booking percentage: {percentage:.2f}%")
```



### Key Observations
- Among restaurants that accept online orders, **3,365 do not offer table booking**, while **362 offer table booking**.
- Among restaurants that do not accept online orders, **2,996 do not offer table booking**, while **382 offer table booking**.
- Approximately **9.71%** of restaurants accepting online orders offer table booking.
- Approximately **11.31%** of restaurants that do not accept online orders offer table booking.

### Conclusion
The analysis shows that most restaurants in both groups do not offer table booking. The proportion offering table booking is slightly higher among restaurants that do not accept online orders. However, this comparison shows an association in the dataset and does not establish that online ordering affects table-booking availability.

## 10. Ratings by Restaurant Type

### Objective
To compare the average ratings of different restaurant types and identify the highest-rated categories.

### Analysis
The `groupby()` function was used to calculate the average rating for each restaurant type. The results were sorted in descending order to identify the highest-rated categories.

To make the comparison more reliable, only restaurant types with at least 10 restaurants were considered. A bar chart was used to visualize the top 10 restaurant types by average rating.



```python
#Ratings by Restaurant type
dataset.groupby('restaurant type')['rate (out of 5)'].mean().sort_values(ascending=False).head(10)
```

```python
#Ratings by Restaurant type - Bar Plot
top_rated = (
    dataset.groupby('restaurant type')['rate (out of 5)']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 5))
plt.bar(top_rated.index, top_rated.values)

plt.title('Top 10 Restaurant Types by Average Rating')
plt.xlabel('Restaurant Type')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

### Key Observations
- **Microbrewery, Casual Dining** has the highest average rating among the selected categories, at approximately **4.15**.
- **Fine Dining** ranks second, with an average rating of approximately **4.12**.
- **Casual Dining, Pub** has an average rating of approximately **4.04**.
- The average ratings of the selected restaurant types range from approximately **3.81 to 4.15**.
- Filtering out categories with fewer than 10 restaurants helps avoid relying on ratings from very small groups.

### Conclusion
The analysis shows that Microbrewery, Casual Dining and Fine Dining have the highest average ratings among the restaurant types examined. These results describe the dataset and do not necessarily indicate that every restaurant in these categories receives a high rating.

## 11. Restaurant Count Analysis

### Objective
To identify the areas with the highest number of restaurants in the dataset.

### Analysis
The `area` column was analyzed using the `value_counts()` function to count the number of restaurants in each area. The results can be sorted in descending order to identify the top 10 areas with the highest restaurant counts.



```python
#Restaurant Count
dataset.groupby('restaurant type')['rate (out of 5)'].agg(
    ['mean', 'count']
).sort_values(by='mean', ascending=False).head(10)
```

```python
#Minimum Restaurant count filter
type_stats = dataset.groupby('restaurant type')['rate (out of 5)'].agg(['mean', 'count'])
filtered_types = type_stats[type_stats['count']>=10]
top_10_reliable = filtered_types.sort_values(by='mean', ascending=False).head(10)
print(top_10_reliable)

plt.figure(figsize=(12, 6))
plt.bar(top_10_reliable.index, top_10_reliable['mean'])
plt.title('Top 10 Restaurant Types by Average Rating')
plt.xlabel('Restaurant Type')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 5)
plt.tight_layout()
plt.show()
```

### Key Observations
The restaurant count analysis helps identify areas with a higher concentration of restaurants. These counts can be used to compare restaurant distribution across different areas.

### Conclusion
This analysis provides an overview of restaurant distribution across areas and helps identify locations with a greater number of restaurants in the dataset.

## 15. Average Cost Analysis

### Objective
To analyze the average cost for two people across different restaurant types and identify the most expensive restaurant categories in the dataset.

### Analysis
The `groupby()` function was used to calculate the average cost for two people and the restaurant count for each restaurant type. To make the comparison more meaningful, only restaurant types with at least 10 restaurants were considered. The results were sorted in descending order of average cost, and the top 10 categories were selected and visualized using a bar chart.



```python
#Cost Analysis

cost_by_type = dataset.groupby('restaurant type')['avg cost (two people)'].agg(['mean', 'count'])

cost_by_type = cost_by_type[cost_by_type['count'] >= 10]

top_costly_types = cost_by_type.sort_values(
    by='mean', ascending=False
).head(10)

print(top_costly_types)


plt.figure(figsize=(12, 6))
plt.bar(top_costly_types.index, top_costly_types['mean'])
plt.title('Top 10 Restaurant Types by Average Cost for Two')
plt.xlabel('Restaurant Type')
plt.ylabel('Average Cost for Two (₹)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

```

### Key Observations
- **Fine Dining** has the highest average cost for two people, at approximately **₹2,784.81**, among the selected restaurant types.
- **Lounge** ranks second, with an average cost of approximately **₹1,771.55**.
- **Microbrewery, Casual Dining** ranks third, with an average cost of approximately **₹1,576.47**.
- **Casual Dining, Cafe** has the lowest average cost among the selected top 10 categories, at approximately **₹1,060**.
- The results show noticeable differences in average cost across restaurant types.

### Conclusion
The analysis indicates that Fine Dining restaurants have the highest average cost for two people among the restaurant types examined, while Lounge and Microbrewery, Casual Dining also have relatively high average costs. These findings describe the cost patterns in the analyzed dataset.
