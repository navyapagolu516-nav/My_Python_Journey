```python
#NumPy copy Vs View
x1= np.arange(10)
x1
```

```python
x2 = x1
x2
```

```python
x2[0] = 10
print(x1)
print(x2)
```

```python
memory = np.shares_memory(x1,x2)
memory
```

```python
# Create a view of x1 and store it in x3.
x3 = x1.view()
x3
```

```python
# Again check memory share between x1 and x3
np.shares_memory(x1, x3)
```

```python
# Now create an array x4 which is copy of x1

x4 = np.copy(x1)
x4[0] = 100
print(x1)
print(x4)
```

```python
#Check memory share between x1 and x4
print(id(x1))
print(id(x4))
np.shares_memory(x1, x4)
```
