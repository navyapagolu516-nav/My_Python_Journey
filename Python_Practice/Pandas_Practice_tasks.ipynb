```python
##Q1. Pandas Series
# marks = [78, 85, 92, 66, 88]
# expected output:
# 0    78
# 1    85
# 2    92
# 3    66
# 4    88

marks = [78, 85, 92, 66, 88]
pd.Series(marks)
```

```python
# Q2. Custom Index
# marks = [78, 85, 92, 66, 88]
# students = ["Navya", "Ramu", "Rohit", "Priya", "Sai"]
# Expected Output:
# Navya     78
# Ramu      85
# Rohit     92
# Priya     66
# Sai       88

marks = [78, 85, 92, 66, 88]
students = ["Navya", "Ramu", "Rohit", "Priya", "Sai"]
pd.Series(marks, index = students)

```

```python
# Q3. Series datatype change
# salary = [25000, 30000, 45000, 50000]
# Expected Output:
# 25000.0
# 30000.0
# 45000.0
# 50000.0

salary = [25000, 30000, 45000, 50000]
pd.Series(salary, dtype= float)
```

```python
# Q4. Dictionary → Series
employee_salary = {
    "E101": 25000,
    "E102": 32000,
    "E103": 41000,
    "E104": 28000
}
pd.Series(employee_salary)  #keys become index numbers
```

```python
#Missing Values Practice
import pandas as pd
import numpy as np

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Arjun", "Meena", np.nan, "Divya", "Kiran"],
    "Department": ["IT", "HR", "IT", np.nan, "HR"],
    "Salary": [30000, 28000, np.nan, 50000, 35000]
}

df21 = pd.DataFrame(data)
df21
```

```python
#Q16. identify Missing values
df21.isnull()
```

```python
#Q17. Null count
df21.isnull().sum()
```

```python
#Q18. Fill Salary missing value
#Q19. Fill missing value for Different columns
df21["Salary"] = df21["Salary"].fillna(df21["Salary"].mean())
df21["Name"] = df21["Name"].fillna("Unknown")
df21["Department"] = df21["Department"].fillna("Not Assigned")
df21
```

```python
#PART-B Dataset creation
import pandas as pd
import numpy as np

employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [
        "Arjun", "Meena", "Rahul", "Divya",
        "Kiran", "Sneha", "Vijay", "Anjali"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "HR"
    ],
    "Employment_Type": [
        "Permanent", "Contract", "Permanent", "Permanent",
        "Contract", "Permanent", "Contract", "Permanent"
    ],
    "Country": [
        "India", "India", "USA", "India",
        "Germany", "USA", "India", "Germany"
    ],
    "Salary": [
        50000, 35000, 60000, 55000,
        38000, 52000, 45000, 42000
    ],
    "Experience": [5, 2, 7, 6, 3, 5, 4, 3]
}

df = pd.DataFrame(employee_data)

print(df)
```
